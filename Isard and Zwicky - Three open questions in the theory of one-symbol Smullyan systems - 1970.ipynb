# Três Questões em Aberto na Teoria de Sistemas Smullyanos de Um Símbolo

Um sistema smullyano é uma quadrupla $S = ( K, V, P, A )$ onde:

- $K$ é um alfabeto finito de símbolos chamados de _constantes_

- $V$ é um alfabeto finito de símbolos chamados de _variáveis_

- $P$ é um alfabeto finito de símbolos chamados de _predicados_

- $A$ é um alfabeto finito de símbolos chamados de _axiomas_ de uma das formas abaixo:
    1. $B_1 \alpha_1$
    2. $B_1 \alpha_1\  \&\  B_2 \alpha_2\  \&\cdots\  \&\  B_n \alpha_n \mapsto B_{n + 1} \alpha_{n + 1}$

    onde $\alpha_i$ são cadeias em $K \cup V$ e $B_i \in P$.

Um teorema de $S$ é qualquer cadeia que ou é um axioma de $S$ ou é passível de obtenção a partir dos axiomas através de um número finito de

1. Substituição uniforme the cadeias não-nulas de $K^*$ (o fecho de Kleene de $K$) por variáveis

2. _modus ponens_ (a implicação nos teoremas)

O fecho de Kleene é o conjunto de todas as cadeias finitas formadas por símbolos de $K$:
$$
    K^* = \bigcup_{k = 0}^\infty K^{( k )},
$$
onde
$$
    K^{( k )}  = \text{conjunto de todas as cadeias de tamanho }k.
$$


Investigar o problema dos quadrados.

Ideias soltas:

Observemos que
$$
    (n + 1)^2 = n^2 + 2n + 1,
$$
ou seja, por recursão, temos que
$$
    ( n + 1 )^2 = \sum_{i = 0}^{n}\{ 2i + 1 \}
$$

Em um sistema smullyiano de um símbolo (OSS) é trivial obter os inteiros com os seguintes axiomas:

- $I|$

- $Ix \to Ix|$

Onde $I$ é o atributo "inteiro" e $|$ é o símbolo. Variáveis serão introduzidas sem cerimônia usando letras romanas minúsculas em itálico. 

Podemos gerar os ímpares adicionando o seguinte axioma:

- $Ix \to Oxx|$

Onde $O$ é o atributo "ímpar".

Consideremos os seguintes axiomas:

- $Sx \to Rxxx|$

Aqui $S$ seria o atributo "quadrado" enquanto $R$ seria o atributo "representação de um quadrado".

Se fosse possível posteriormente levar de $Rxxx|$ para o quadrado de $|x| + 1$, teríamos resolvido o problema.


Ou, talvez,

- $Ix\ \&\ Sy \to Syxx|$

O problema aqui é que não conseguimos associar $Sy$ ao quadrado de $|x|$, o que faz com que o axioma não resulte no que desejamos.
